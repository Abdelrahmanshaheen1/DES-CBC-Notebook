```python
#DES ENCRYPT
def permute(block, table):
    """Rearranges bits in 'block' based on 'table'."""
    permuted_block = ''.join(block[pos - 1] for pos in table)
    return permuted_block

def xor(bits1, bits2):
    return ''.join('1' if b1 != b2 else '0' for b1, b2 in zip(bits1, bits2))

def apply_sbox_substitution(bits):
    """Applies S-Box substitution to the 48-bit input."""
    output_bits = ""
    for i in range(8):
        six_bits = bits[i*6:(i+1)*6]
        row = int(six_bits[0] + six_bits[-1], 2)
        col = int(six_bits[1:5], 2)
        sbox_value = S_BOXES[i][row][col]
        output_bits += format(sbox_value, '04b')  # Convert to 4-bit binary
    return output_bits

def apply_pbox_permutation(bits):
    """Applies P-Box permutation to the 32-bit input."""
    return permute(bits, P_BOX)

def apply_ipbox_permutation(bits):
    return permute(bits, IP)


def des_encrypt_64bit_block(block, round_keys, verbose=True):
    """Encrypts a 64-bit block using DES over 16 rounds."""
    block=apply_ipbox_permutation(block)
    if verbose: print("block after intial permintation"+block)
    L, R = block[:32], block[32:]
    if verbose: print("L0 = ",L,"R0 = ",R)
    for i in range(16):
        expanded_R = permute(R,E)
        mixed_R = xor(expanded_R, round_keys[i])
        substituted_R = apply_sbox_substitution(mixed_R) # (Substitution and permutation would go here)
        permuted_R = apply_pbox_permutation(substituted_R)
        L, R = R, xor(L, permuted_R)
        if verbose: print(f"Round {i+1}: L = {L}, R = {R}")

    final_block = permute(R+ L, FP)
    return final_block

# Test input block and first round key
test_round_keys= ["000110110000001011101111111111000111000001110010"] * 16  # Replicate key for all 16 rounds

# Run the function with the test input
block="0000000100100011010001010110011110001001101010111100110111101111"
cipher_output = des_encrypt_64bit_block(block, test_round_keys)
print("Ciphertext after 16 rounds:", cipher_output)


```

## Extension: real key schedule, decryption and tests

The cells above use a fixed placeholder round key. This extension adds DES's real **key schedule**
(PC-1, the per-round left rotations, PC-2) so that a single 64-bit key produces the 16 different
round keys, plus decryption and CBC decryption, and checks everything against the standard DES test
vector (key `133457799BBCDFF1`, plaintext `0123456789ABCDEF`, ciphertext `85E813540F0AB405`).

Still educational only: single DES has a 56-bit key and is broken, the demo IV below is all zeros,
and the zero padding is not reversible in general (real systems use PKCS#7 padding and a random IV).

```python
# --- Key schedule -------------------------------------------------------------
PC1 = [57, 49, 41, 33, 25, 17, 9, 1, 58, 50, 42, 34, 26, 18,
       10, 2, 59, 51, 43, 35, 27, 19, 11, 3, 60, 52, 44, 36,
       63, 55, 47, 39, 31, 23, 15, 7, 62, 54, 46, 38, 30, 22,
       14, 6, 61, 53, 45, 37, 29, 21, 13, 5, 28, 20, 12, 4]          # 64 -> 56 bits

PC2 = [14, 17, 11, 24, 1, 5, 3, 28, 15, 6, 21, 10,
       23, 19, 12, 4, 26, 8, 16, 7, 27, 20, 13, 2,
       41, 52, 31, 37, 47, 55, 30, 40, 51, 45, 33, 48,
       44, 49, 39, 56, 34, 53, 46, 42, 50, 36, 29, 32]               # 56 -> 48 bits

SHIFTS = [1, 1, 2, 2, 2, 2, 2, 2, 1, 2, 2, 2, 2, 2, 2, 1]            # left rotations per round


def hex_to_bits(h):
    return ''.join(format(int(ch, 16), '04b') for ch in h)

def bits_to_hex(b):
    return ''.join(format(int(b[i:i+4], 2), 'X') for i in range(0, len(b), 4))

def left_rotate(bits, n):
    return bits[n:] + bits[:n]

def generate_round_keys(key_bits):
    """Derive the 16 round keys (48 bits each) from a 64-bit key."""
    assert len(key_bits) == 64
    permuted = permute(key_bits, PC1)            # drop the 8 parity bits
    C, D = permuted[:28], permuted[28:]
    keys = []
    for shift in SHIFTS:
        C, D = left_rotate(C, shift), left_rotate(D, shift)
        keys.append(permute(C + D, PC2))
    return keys
```

```python
# --- Decryption ---------------------------------------------------------------
def des_decrypt_64bit_block(block, round_keys, verbose=False):
    """DES decryption is encryption with the round keys in reverse order."""
    return des_encrypt_64bit_block(block, round_keys[::-1], verbose)

def bits_to_text(bits):
    data = bytes(int(bits[i:i+8], 2) for i in range(0, len(bits), 8))
    return data.decode('utf-8', errors='replace')

def des_decrypt_cbc(cipher_bits, round_keys, iv, verbose=False):
    """Inverse of des_encrypt_cbc. Removes the leading zero padding of the last block
    (only safe for text that does not itself begin with NUL characters)."""
    blocks = [cipher_bits[i:i+64] for i in range(0, len(cipher_bits), 64)]
    previous = iv
    plain_bits = ''
    for block in blocks:
        plain_bits += xor(des_decrypt_64bit_block(block, round_keys, verbose), previous)
        previous = block
    return bits_to_text(plain_bits).replace('\x00', '')
```

```python
# --- Tests --------------------------------------------------------------------
KEY = hex_to_bits("133457799BBCDFF1")
PLAIN = hex_to_bits("0123456789ABCDEF")
EXPECTED = hex_to_bits("85E813540F0AB405")

keys = generate_round_keys(KEY)
assert len(keys) == 16 and all(len(k) == 48 for k in keys)
assert keys[0] == "000110110000001011101111111111000111000001110010"    # K1 from the classic worked example
assert keys[15] == "110010110011110110001011000011100001011111110101"   # K16 from the same example
assert len(set(keys)) == 16                                              # round keys now differ

cipher = des_encrypt_64bit_block(PLAIN, keys, verbose=False)
assert cipher == EXPECTED, bits_to_hex(cipher)
assert des_decrypt_64bit_block(cipher, keys) == PLAIN

# CBC round trip on a multi-block message
IV = hex_to_bits("0000000000000000")
message = "Attack at dawn! Attack at dawn! Attack at dawn!"
encrypted = des_encrypt_cbc(message, keys, IV, verbose=False)
assert des_decrypt_cbc(encrypted, keys, IV) == message

# Cross-check against a reference implementation when one is installed
try:
    from Crypto.Cipher import DES as RefDES
    import os
    for _ in range(50):
        k, p, iv = os.urandom(8), os.urandom(8), os.urandom(8)
        mine = bits_to_hex(des_encrypt_64bit_block(hex_to_bits(p.hex()), generate_round_keys(hex_to_bits(k.hex())), verbose=False))
        assert mine == RefDES.new(k, RefDES.MODE_ECB).encrypt(p).hex().upper()
    text = "CBC cross-check with a reference library, 3 blocks."
    k = os.urandom(8); iv = os.urandom(8)
    # the notebook left-pads with zeros, so compare on a message that is already a multiple of 8 bytes
    text8 = text + " " * ((-len(text)) % 8)
    ref = RefDES.new(k, RefDES.MODE_CBC, iv).encrypt(text8.encode())
    mine = bits_to_hex(des_encrypt_cbc(text8, generate_round_keys(hex_to_bits(k.hex())), hex_to_bits(iv.hex()), verbose=False))
    assert mine == ref.hex().upper()
    print("Matches pycryptodome on 50 random ECB blocks and one CBC message.")
except ImportError:
    print("pycryptodome not installed: skipped the reference cross-check.")

print("Known-answer vector OK:", bits_to_hex(cipher))
print("All checks passed.")
```
